```python
import pandas as pd
import numpy as np
from datetime import datetime, timedelta

# Configuration for the 31-column Production Schema
N_STAYS = 1000
START_DATE = datetime(2026, 3, 29, 8, 0)

def generate_updated_csv():
    rows = []
    complaint_map = {
        "Cardiac": ["chest pain", "shortness of breath", "palpitations"],
        "Trauma": ["s/p fall", "hip pain", "head injury"],
        "Minor": ["refill request", "skin rash"]
    }

    for i in range(N_STAYS):
        stay_id = 3000000 + i
        arrival_time = START_DATE + timedelta(minutes=i * 15)
        
        # Injected Surge Logic (4 PM - 8 PM)
        multiplier = 3.5 if (16 <= arrival_time.hour <= 20) else 1.0

        # Multi-Label Complaints
        cat = np.random.choice(list(complaint_map.keys()))
        n_tokens = np.random.randint(2, 4) if cat != "Minor" else 1
        complaints = ", ".join(np.random.choice(complaint_map[cat], n_tokens, replace=False))
        
        # 31-Column Row Generation
        rows.append({
            'stay_id': stay_id, 'subject_id': 10000+i, 'hadm_id': 20000+i,
            'timestamps': arrival_time + timedelta(minutes=np.random.exponential(45 * multiplier)),
            'activity': 'triage in the ed', 'gender': 'F', 'race': 'WHITE',
            'arrival_transport': 'AMBULANCE', 'disposition': 'ADMITTED',
            'seq_num': 1, 'icd_code': 'R07.9', 'icd_version': 10, 'icd_title': 'Chest pain',
            'temperature': 98.6, 'heartrate': 102, 'resprate': 18, 'o2sat': 98,
            'sbp': 135, 'dbp': 85, 'pain': 8, 'acuity': 2, 'chiefcomplaint': complaints,
            'rhythm': 'SR', 'name': 'Saline', 'gsn': '00123', 'ndc': '00409-6012-03',
            'etc_rn': 1, 'etccode': 101, 'etcdescription': 'Electrolytes', 'med_rn': 1, 'gsn_rn': 1
        })

    df = pd.DataFrame(rows)
    df.to_csv('ed_production_sim_1k.csv', index=False)
    print("New high-fidelity CSV generated: ed_production_sim_1k.csv")

generate_updated_csv()
```
